```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Correto = pd.read_csv('Correto.csv',delimiter=';')
BruteForce = pd.read_csv('BruteForce.csv',delimiter=';')
Dynamic = pd.read_csv('Dynamic.csv',delimiter=';')
Greedy = pd.read_csv('Greedy.csv',delimiter=';')
```

```python
# Criando o gráfico
plt.figure(figsize=(8, 6))

# Plotando cada DataFrame como uma linha
plt.plot(Correto["vertices"], Correto["qualidade"], marker="o", linestyle="-", label="Correto", alpha=1)
plt.plot(BruteForce["vertices"], BruteForce["qualidade"], marker=".", linestyle="-", label="BruteForce", alpha=1)
plt.plot(Dynamic["vertices"], Dynamic["qualidade"], marker="v", linestyle="-", label="Dynamic", alpha=1)
plt.plot(Greedy["vertices"], Greedy["qualidade"], marker="x", linestyle="-", label="Greedy", alpha=1)

# Adicionando rótulos e título
plt.xlabel("vertices")
plt.ylabel("Qualidade")
plt.legend(title="DataFrames")
plt.title("Comparação da Qualidade com o Aumento do vertices")

# Exibir o gráfico
plt.show()
```
